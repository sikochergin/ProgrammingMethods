```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("sortingtime.csv")

df
```

```python
plt.figure(figsize=(12, 7))

slow_alg = df[df["algorithm"] != "heapSort"][df["algorithm"] != "stdSort"]

for algorithm in slow_alg["algorithm"].unique():
    data = slow_alg[slow_alg["algorithm"] == algorithm]

    plt.plot(
        data["size"],
        data["time_microseconds"],
        marker="o",
        linewidth=2,
        label=slow_alg.get(algorithm, algorithm)
    )

plt.title("Зависимость времени сортировки от размера выборки", fontsize=16)
plt.xlabel("Размер выборки", fontsize=12)
plt.ylabel("Время сортировки, мкс", fontsize=12)

plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.tight_layout()

plt.savefig("sort_time_slow.png", dpi=300)
plt.show()
```
